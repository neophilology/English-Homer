# PDF to Text Application

```python
# TO Update
translator = "Fagles"
pdf_file = f'/Users/debr/Documents/MA-ELC/ThESis/Odyssey Translations/Odyssey_{translator}.pdf'
txt_file = f"/Users/debr/odysseys_en/raw_txts/Odyssey_{translator}.txt"  
```

```python
import pdfplumber

# Function to extract text from PDF
def extract_text_from_pdf(pdf_path, txt_path):
    with pdfplumber.open(pdf_path) as pdf:
        text = []
        for page in pdf.pages:
            # Extract text from the page while preserving line breaks
            page_text = page.extract_text()
            if page_text:  # Check if text extraction was successful
                text.append(page_text)
        
        # Join all pages with double line breaks
        full_text = "\n\n".join(text)
        
        # Write to text file
        with open(txt_path, "w", encoding="utf-8") as f:
            f.write(full_text)
        
        print(f"Text extracted and saved to {txt_path}")

# Conversion

extract_text_from_pdf(pdf_file, txt_file)
```
